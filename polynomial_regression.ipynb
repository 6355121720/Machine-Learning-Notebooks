```python
import numpy as np
import numpy.random as rd
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
```

```python
x=6*rd.rand(200,1)+1
y=17*(x**2)+3*x-12+90*rd.rand(200,1)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
sns.scatterplot(x=x.ravel(),y=y.ravel())
plt.show()
```

```python
poly=PolynomialFeatures(degree=2)

x_train_trans=poly.fit_transform(x_train)
x_test_trans=poly.transform(x_test)

x_train_trans
```

```python
m1=LinearRegression()
m1.fit(x_train_trans,y_train)
r2_score(y_test,m1.predict(x_test_trans))
```

```python
m1.coef_
```

```python
m1.intercept_
```

```python
sns.lineplot(x=x_test.flatten(),y=m1.predict(x_test_trans).flatten(),color='r',label='Model Line')
sns.scatterplot(x=x_test.flatten(),y=y_test.flatten(),label='Test Datapoints')
plt.legend()
plt.show()
```
