```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,SGDRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import seaborn as sns
```

```python
x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=2)
```

```python
m1=LinearRegression()
m1.fit(x_train,y_train)
r2_score(y_test,m1.predict(x_test))
```

```python
class my_SGD:
    def __init__(self):
        self.lr=0.1
        self.epochs=100
        self.coef_=None
        self.intercept_=0
    def fit(self,x,y):
        self.coef_=np.ones(x.shape[1])
        for i in range(self.epochs):
            for j in range(x.shape[0]):
                idx=np.random.randint(0,x.shape[0])
                y_hat=np.dot(x[idx],self.coef_)+self.intercept_
                slope_i=-2*(y[idx]-y_hat)
                slope_c=-2*np.dot((y[idx]-y_hat),x[idx])

                self.coef_=self.coef_-(self.lr*slope_c)
                self.intercept_=self.intercept_-(self.lr*slope_i)
        print(self.intercept_,self.coef_)
    def predict(self,x):
        return np.dot(x,self.coef_)+self.intercept_
```

```python
m2=my_SGD()
m2.fit(x_train,y_train)
r2_score(y_test,m2.predict(x_test))
```
