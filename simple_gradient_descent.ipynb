```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_iris,make_regression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
data=make_regression(n_samples=20,n_features=1,n_informative=1,n_targets=1,noise=10,random_state=2)
x=data[0]
y=data[1]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=2)
```

```python
sns.scatterplot(x=x.flatten(),y=y.flatten())
plt.show()
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
r2_score(y_test,model.predict(x_test))
```

Gradient Descent

```python
#if we know value of any one parameter we can find another one.

lr=0.01
epochs=100
b=1
m=42.93
for i in range(epochs):
    slope=-2*np.sum(y_train.ravel()-m*x_train.ravel()-b)
    b=b-lr*slope
print(b,m)
```

```python
# case when we dont know anything

lr=0.01
epochs=1000
b=1
m=1
for i in range(epochs):
    slope_b=-2*np.sum(y_train.ravel()-m*x_train.ravel()-b)
    slope_m=-2*np.sum((y_train.ravel()-m*x_train.ravel()-b)*x_train.ravel())

    b=b-lr*slope_b
    m=m-lr*slope_m

print(b,m)
```
