```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.ensemble import BaggingClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
```

```python
x,y=make_classification(n_samples=1000,n_features=10,n_informative=6,random_state=2)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
x,y
```

```python
m1=DecisionTreeClassifier()
m1.fit(x_train,y_train)
accuracy_score(y_test,m1.predict(x_test))
```

Bagging

```python
m2=BaggingClassifier(estimator=DecisionTreeClassifier(),n_estimators=100,max_samples=0.8,bootstrap=True,random_state=2 )
m2.fit(x_train,y_train)
accuracy_score(y_test,m2.predict(x_test))
```

Pasting

```python
m3=BaggingClassifier(estimator=DecisionTreeClassifier(),n_estimators=100,max_samples=0.7,bootstrap=False,random_state=2)
m3.fit(x_train,y_train)
accuracy_score(y_test,m3.predict(x_test))
```

Random Subspaces

```python
m4=BaggingClassifier(estimator=DecisionTreeClassifier(),n_estimators=100,max_samples=1,bootstrap=False,
                     max_features=0.5,bootstrap_features=True,random_state=2)
m4.fit(x_train,y_train)
accuracy_score(y_test,m4.predict(x_test))
```

Random Patches

```python
m5=BaggingClassifier(estimator=DecisionTreeClassifier(),n_estimators=100,max_samples=0.8,bootstrap=True,
                     max_features=0.5,bootstrap_features=True,random_state=2)
m5.fit(x_train,y_train)
accuracy_score(y_test,m5.predict(x_test))
```
