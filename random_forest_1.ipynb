```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
x,y=make_classification(n_samples=100,n_features=10,n_informative=6,random_state=2)
```

```python
def rowsampling(x,y,pr,state=2):
    np.random.seed(state)
    idx=np.random.choice(np.arange(x.shape[0]),size=(int(x.shape[0]*pr),),replace=True)
    x1=x[idx]
    y1=y[idx]
    return x1,y1
```

```python
x_train,x_test,y_train,y_test=train_test_split(rowsampling(x,y,0.6,4)[0],rowsampling(x,y,0.6,4)[1],test_size=0.2,random_state=2)
```

```python
m1=DecisionTreeClassifier()
m1.fit(x_train,y_train)
accuracy_score(y_test,m1.predict(x_test))
```

```python
plot_tree(m1)
```
