```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('datasets/iris.csv',names=['col'+str(i) for i in range(1,4)]+['target'])
data.head()
```

```python
data.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss=StandardScaler()
data=ss.fit_transform(np.array(data.iloc[:,:-1]).T)
```

```python
np.set_printoptions(formatter={'float': '{: 0.3f}'.format})
```

```python
data
```

```python
cov_matrix=np.cov(data)
```

```python
cov_matrix
```

```python
eigvalues, eigvectors = np.linalg.eig(cov_matrix)
```

```python
eigvalues
```

```python
eigvectors
```

```python
pc1 = np.dot(data.T,eigvectors[0])
pc2 = np.dot(data.T,eigvectors[1])
pc3 = np.dot(data.T,eigvectors[2])
```

## PCA

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
pca=PCA(n_components=3)
```

```python
data=pd.read_csv('datasets/iris.csv',names=['col'+str(i) for i in range(1,5)]+['target'])
```

```python
data.head()
```

```python
X=np.array(data.iloc[:,:-1])
y=np.array(data.iloc[:,-1])
```

```python
ss=StandardScaler()
X=ss.fit_transform(X)
X,y
```

```python
X=pca.fit_transform(X)
```

```python
X
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
y=encoder.fit_transform(y)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier(n_neighbors=4)
knn.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,knn.predict(X_test))
```

# Original Data

```python
data=pd.read_csv('datasets/iris.csv',names=['col'+str(i) for i in range(1,5)]+['target'])
```

```python
X=data.iloc[:,:-1]
y=data.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
knn2=KNeighborsClassifier(n_neighbors=4)
knn2.fit(X_train,y_train)
```

```python
accuracy_score(y_test,knn2.predict(X_test))
```
