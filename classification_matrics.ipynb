```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score,accuracy_score,confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv('datasets/heart_attack.csv')
x=np.array(data.iloc[:,0:data.shape[1]-1])
np.set_printoptions(suppress=True,precision=2) # suppresses scientific notation into normal floating point notation
print(x)
y=data.iloc[:,data.shape[1]-1]
print(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
m1=LogisticRegression()
m1.fit(x_train,y_train)
m2=DecisionTreeClassifier()
m2.fit(x_train,y_train)
print('LR Accuracy : ',accuracy_score(y_test,m1.predict(x_test)))
print('DT Accuracy : ',accuracy_score(y_test,m2.predict(x_test)))
```

```python
print('LR confusion matrix : ',confusion_matrix(y_test,m1.predict(x_test)),sep='\n',end='\n\n')
print('DT confustion matrix : ',confusion_matrix(y_test,m2.predict(x_test)),sep='\n')
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
```

```python
data=pd.read_csv('datasets/heart_attack.csv')
x=np.array(data.iloc[:,0:data.shape[1]-1])
np.set_printoptions(suppress=True,precision=2) # suppresses scientific notation into normal floating point notation
print(x)
y=data.iloc[:,data.shape[1]-1]
print(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
m2=DecisionTreeClassifier()
m2.fit(x_train,y_train)
print('DT confustion matrix : ',confusion_matrix(y_test,m2.predict(x_test)),sep='\n')
```

```python
print('Precision score : ',precision_score(y_test,m2.predict(x_test)),precision_score(y_test,m2.predict(x_test),average=None))
print('Recall score : ',recall_score(y_test,m2.predict(x_test)),recall_score(y_test,m2.predict(x_test),average=None))
print('F1 score : ',f1_score(y_test,m2.predict(x_test)),f1_score(y_test,m2.predict(x_test),average=None))
```
