```python
import numpy as np
import pandas as pd
from mlxtend.plotting import plot_decision_regions
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import warnings
warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv('datasets/iris.csv',header=None)
e=LabelEncoder()
data[4]=e.fit_transform(data[4])
data=data[data[4]!=0]
x=np.array(data.iloc[:,:-1])
y=np.array(data.iloc[:,-1])
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=2)
x,y
```

```python
m1=AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=1),n_estimators=100,random_state=2)
m1.fit(x_train,y_train)
accuracy_score(y_test,m1.predict(x_test))
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param={'n_estimators':[25,50,100],'learning_rate':[0.2,0.5,1],'algorithm':['SAMME','SAMME.R']}
m2=GridSearchCV(AdaBoostClassifier(),param_grid=param,cv=5,scoring='accuracy')
m2.fit(x,y)
m2.best_score_,m2.best_params_
```
