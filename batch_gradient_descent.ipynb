```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import seaborn as sns
```

```python
x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=2)
```

```python
m1=LinearRegression()
m1.fit(x_train,y_train)
r2_score(y_test,m1.predict(x_test))
```

By GD

```python
class my_GD:
    def __init__(self):
        self.lr=0.1
        self.epochs=1000
        self.coef_=None
        self.intercept_=0
    def fit(self,x,y):
        self.coef_=np.ones(x.shape[1])
        for i in range(self.epochs):
            y_hat=np.dot(x,self.coef_)+self.intercept_
            slopes_coef=-2*np.dot((y-y_hat).T,x)/x.shape[1]
            slope_intercept=-2*np.mean(y-y_hat)/x.shape[1]
            self.coef_=self.coef_-(self.lr*slopes_coef)
            self.intercept_=self.intercept_-(self.lr*slope_intercept)
        print(self.coef_,self.intercept_)
    def predict(self,x):
        return np.dot(x,self.coef_)+self.intercept_
```

```python
m2=my_GD()
m2.fit(x_train,y_train)
```

```python
r2_score(y_test,m2.predict(x_test))
```
