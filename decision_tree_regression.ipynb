```python
from sklearn.tree import DecisionTreeRegressor
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
data=pd.read_csv('datasets/housing.csv',header=None)
data=pd.DataFrame([spl(i) for i in data[0]])
data.drop(columns=[3])
x=data.iloc[:,:12]
y=data.iloc[:,12]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=0)
x,y
```

```python
def spl(s):
    l=[]
    temp=''
    for i in s:
        if(i==' '):
            if bool(temp):
                l.append(temp)
            temp=''
        else:
            temp=temp+i
    return l
```

```python
m1=DecisionTreeRegressor(max_depth=3,criterion='squared_error')
m1.fit(x_train,y_train)
r2_score(y_test,m1.predict(x_test)),m1.predict(x_test.iloc[:10,:]),y_test.iloc[:10]
```
