```python
import pandas as pd
import numpy as np
import seaborn as sns 
from sklearn.datasets import load_diabetes
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
import random
```

```python
x,y=load_diabetes(return_X_y=True)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=2)
```

```python
m1=SGDRegressor(max_iter=1000,learning_rate='constant',eta0=0.01)
m1.fit(x_train,y_train)
r2_score(y_test,m1.predict(x_test))
```

```python
class my_MBGD:
    def __init__(self):
        self.lr=0.5
        self.epochs=400
        self.coef_=None
        self.intercept_=0
        self.b_size=100
    def fit(self,x,y):
        self.coef_=np.ones(x.shape[1])
        for i in range(self.epochs):
            for j in range(int(x.shape[0]/self.b_size)):
                idx=random.sample(range(x.shape[0]),self.b_size)
                y_bar=np.dot(x[idx],self.coef_)+self.intercept_
                slope_i=-2*np.mean(y[idx]-y_bar)
                slope_c=-2*np.dot((y[idx]-y_bar).T,x[idx])/x.shape[0]

                self.coef_=self.coef_-(self.lr*slope_c)
                self.intercept_=self.intercept_-(self.lr*slope_i)
        print(self.coef_,self.intercept_)
    def predict(self,x):
        return np.dot(x,self.coef_)+self.intercept_
```

```python
m2=my_MBGD()
m2.fit(x_train,y_train)
r2_score(y_test,m2.predict(x_test))
```
