```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv('datasets/iris.csv',header=None)
data.drop(columns=[1,3],inplace=True)
x=np.array(data.iloc[:,:2])
y=data.iloc[:,2]
e=LabelEncoder()
y=e.fit_transform(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
x,y
```

```python
tree=DecisionTreeClassifier(max_depth=4)
tree.fit(x_train,y_train)
accuracy_score(y_test,tree.predict(x_test))
```

```python
a=np.linspace(x_test[:,0].min(),x_test[:,0].max(),100)
b=np.linspace(x_test[:,1].min(),x_test[:,1].max(),100)
xx,yy=np.meshgrid(a,b)
y_pred=tree.predict(np.array([xx.ravel(),yy.ravel()]).T).reshape(xx.shape)
plt.contourf(xx,yy,y_pred,alpha=0.9)
sns.scatterplot(x=x_train[:,0],y=x_train[:,1],c=y_train)
plt.colorbar()
plt.show()
```

Another Example

```python
data=pd.read_csv('datasets/Social_Network_Ads.csv')
data
x=data.iloc[:,2:4].values
y=data.iloc[:,4].values
x,y
```

```python
sns.scatterplot(x=x[:,0],y=x[:,1],c=y)
plt.show()
```

```python
def DTC(max_depth):
    m1=DecisionTreeClassifier(max_depth=max_depth)
    m1.fit(x,y)
    a=np.linspace(x[:,0].min()-1,x[:,0].max()+1,100)
    b=np.linspace(x[:,1].min()-1,x[:,1].max()+1,100)
    xx,yy=np.meshgrid(a,b)
    input_array=np.array([xx.ravel(),yy.ravel()]).T
    output_array=m1.predict(input_array)
    plt.contourf(xx,yy,output_array.reshape(xx.shape),cmap='OrRd')
    sns.scatterplot(x=x[:,0],y=x[:,1],c=y)
    plt.colorbar()
    plt.show()
```
