```python
import numpy as np
import pandas as pd
```

```python
df3=pd.DataFrame({'name':['n1','n2','n3','n2','n1'],'score':[2,4,8,5,3]})
df3=pd.DataFrame(np.array([['n1','n2','n3','n2','n1'],[2,4,8,5,3]]).T,columns=['name','score'])
df3
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder()
encoded=pd.DataFrame(ohe.fit_transform(df3[['name']]).toarray(),columns=ohe.get_feature_names_out())
```

```python
pd.concat([df3,encoded],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
encoded=pd.DataFrame(le.fit_transform(df3[['name']]),columns=['name'])
encoded
```

```python
pd.concat([df3,encoded],axis=1)
```

```python
df4=pd.DataFrame({'score':['small','medium','small','large','medium']})
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=[['small','medium','large']])
encoded=pd.DataFrame(oe.fit_transform(df4[['score']]))
encoded
```

```python
pd.concat([df4,encoded],axis=1)
```
